```kotlin
//task 1: Необходимо написать функцию, которая находит целую часть квадратного корня числа (округление вниз).
fun find_sqrt(number: Int): Int{
    //границы поиска корня в общем случае от 1 до n/2 для натуральных чисел
    var left:Int = 1
    var right:Int = number/2 + 1   //+1 чтобы на маленьких числах не отбрасывало искомое значение при округлении вниз
    var middle:Int = left + (right - left)/2

    while (right-left>1) 
    {   
        if (middle*middle == number) return middle
        if (middle*middle > number) right = middle
        else if (middle*middle < number) left = middle
        middle = left + (right - left)/2
    }
    return left

}

fun main(){
    val numbers = listOf(2,3,4,5,8,9,10,15,17)

    println(find_sqrt(4))
    numbers.forEach { number -> println("$number >= ${find_sqrt(number)}^2") }
}

main()

// алгоритм работает за O(log N), так область поиска каждый раз сужается вдвое, максимальное количество возможных целочисленных делений на 2 числа N равно log N
```

```kotlin
//task 2 Как быстро можно сделать N копий документа, если есть два ксерокса, каждый копирует с разной скоростью — x и y минут за копию


fun fastCopy(N:Int, x: Int, y:Int):Int{
    var middle=0
    var left = 0
    var right = N*min(x,y) // время, за которое напечатаем быстрым принтером
    while (right>left){
        middle = left + (right-left)/2
        if (isTimeCorrect(middle,N, x,y)) right = middle else left = middle+1 
        //если время корректно, пытаемся уменьшить- двигаемся влево. если нет, увеличиваем,двигаясь вправо
    }
    return left
}

fun isTimeCorrect(time:Int, N:Int, x: Int, y: Int):Boolean{
    return time/x + time/y >= N //time/x - сколько за это время напечатает первый, time/y - сколько второй, проверяем достаточно ли экземпляров успеем напечатать
}

fun main(){
    println(fastCopy(3, 2, 1))
    println(fastCopy(5, 2, 1))
    println(fastCopy(4, 2, 1))
    println(fastCopy(4, 2, 2))
}
main()


```
