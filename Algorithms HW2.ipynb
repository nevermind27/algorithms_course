```kotlin
// task 1
class Node(var valuee: Int){
    var next: Node? = null
}

class customList{
    var head: Node? = null
    fun addNode(valuee: Int){
        val newNode = Node(valuee) 
        if (head == null) head = newNode
        else
        {
            var currentNode = head
            while (currentNode?.next!=null){
                currentNode = currentNode.next
            }
            currentNode?.next = newNode
        }
    }
    fun printList()
    {
        var currentNode = head
        while (currentNode?.next!=null){
            print(currentNode?.valuee)
            currentNode = currentNode?.next
        }
        print(currentNode?.valuee)
    }
}

// непосредственно функция разворота списка
fun reverseList(sourceList:customList)
{

    var prev:Node? = null   
    var current:Node? = sourceList.head // начинаю с головы списка, копирую ее в current
    var next:Node? = sourceList.head?.next 
    while (next!=null){
        next = current?.next  //сохраняю копию указателя на следующий элемент
        current?.next = prev //меняю указатель в текущем элементе: вместо следующего ставлю предыдущий
        prev = current //двигаю указатели
        if (next!=null) current = next
    }
    sourceList.head = current
}

fun main()
{
    // пример использования
    val newList = customList()
    newList.addNode(9)
    newList.addNode(8)
    newList.addNode(7)

    newList.printList()
    reverseList(newList)
    println()
    newList.printList()
}
main()
```

```kotlin
// task 2

// классы для списка и узлов использую из первого задания
// для четного количества элементов возвращаем первый из пары средних
fun findCentral(list: customList):Int?{
    var fastPointer = list.head?.next
    var slowPointer = list.head
    var stepCounter = 0
    while (fastPointer?.next!=null){
        if (stepCounter%2==0) slowPointer=slowPointer?.next //медленный двигается в 2 раза медленнее. Если быстрый дошел до конца списка,то медленный оказывается ровно на середине
        fastPointer = fastPointer.next
        stepCounter++
    }
    return slowPointer?.valuee
}

fun main(){
    val newList = customList()

    newList.addNode(9)
    //для единичного элемента
    println(findCentral(newList))

    newList.addNode(8)
    newList.addNode(7)
    newList.addNode(6)
    newList.addNode(5)
    newList.addNode(4)
    //вывод для четного кол-ва
    println(findCentral(newList))

    newList.addNode(3)
    //для нечетного
    println(findCentral(newList))
    
}
main()
```

```kotlin
//task 7
/*
Дан отсортированный массив.
Необходимо удалить дубликаты in-place так, чтобы каждый элемент встречался только один раз
*/


fun <T> removeDuplicates(arr: Array<T>): Int{
    var first:Int = 0
    var second:Int = 0
    while (second<arr.size-1 ){
        if (arr[second] == arr[second+1]) second++
        else {
            first ++
            second++
            arr[first] = arr[second]
        }
        
    }
    return first //номер индекса first - последний измененный элемент, то есть новая длина фактического массива = fisrt + 1, то что после него first - считаем мусором
    // как раз с помощью записи вместо предыдущих дублирующих элементов позволяет не аллоцировать новый массив
}

fun main(){
    val a = arrayOf(2, 2, 2, 3, 3, 5, 5, 5)
    
    println(removeDuplicates(a))
    println(a.contentToString())

    val b = arrayOf( 1, 2 , 3) // проверка на массиве без дублей
    println(removeDuplicates(b))
    println(b.contentToString())


}
main()
```
